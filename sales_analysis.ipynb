```python
import pandas as pd
```

```python
data=pd.read_csv('sales_data_sample.csv', encoding='latin1')
print(data.head())
```

# DATA CLEANING

```python
data
```

```python
data=data.drop_duplicates()
data
```

```python
data = data.drop(columns=["QTR_ID", "ADDRESSLINE2", "MSRP", "STATE", "TERRITORY", "DEALSIZE","POSTALCODE","CONTACTLASTNAME","CONTACTFIRSTNAME"], errors='ignore')
data
```

```python
data["CUSTOMERNAME"]=data["CUSTOMERNAME"].str.rstrip(".com")
# remove all the mentionned things from the order date
data["ORDERDATE"]=data["ORDERDATE"].str.strip("0:00")
data["ADDRESSLINE1"]=data["ADDRESSLINE1"].str.strip("C/")
```

```python
data
```

```python
# Make all the phone numbers with the same format
data["PHONE"]=data["PHONE"].str.replace('[^a-zA-Z0-9]','')
data
```

```python
data["PHONE"]=data["PHONE"].str.replace("nan","")
data["PHONE"]=data["PHONE"].str.replace("NAN","")
data
```

```python
print(data.isnull().sum())
```

```python
# having customerid
# filtered_data =data[data['CUSTOMERNAME'] == 'Toys4GrownUps']
# print(filtered_data) 

# Generate unique customer IDs
data['CUSTOMER_ID'] = pd.factorize(data['CUSTOMERNAME'])[0]

print(data)
```

# AGGREGATE DATA

####   MONTHLY SALES TRENDS

```python
monthly_sales = data.groupby('MONTH_ID')['SALES'].sum().reset_index()

print(monthly_sales)
```

####   CUSTOMER PURCHASE BEHAVIOR

##### Average purchase value per customer

```python

# total_purchase_per_customer = data.groupby('CUSTOMER_ID')['SALES'].sum()

# # Calculate the number of customers
# num_customers = data['CUSTOMER_ID'].nunique()

# # Calculate average purchase value per customer
# average_purchase_value = total_purchase_per_customer.sum() / num_customers


# Group by CUSTOMER_ID and calculate the mean sales for each customer
average_sales_per_customer = data.groupby('CUSTOMER_ID')['SALES'].mean().reset_index()

# Rename the columns for clarity (optional)
average_sales_per_customer.columns = ['CUSTOMER_ID', 'SALES']

# Display the DataFrame
print(f'Average purchase value per customer:\n{average_sales_per_customer}')

```

##### Most purchased products

```python
most_purchased = data.groupby('PRODUCTLINE')['QUANTITYORDERED'].sum().reset_index()

# Sort by quantity in descending order
most_purchased_sorted = most_purchased.sort_values(by='QUANTITYORDERED', ascending=False)

# Print the most purchased products
print(most_purchased_sorted)
```

###  PEAK SEASON

##### MONTHS WITH THE HIGHEST SALES

```python
highest_sales= data.groupby('MONTH_ID')['SALES'].sum().reset_index()

highest=highest_sales.sort_values(by='SALES', ascending=False)
print(highest)
```

# Visualize Insights

### Line plot for monthly sales trends.

```python
import matplotlib.pyplot as plt



# Plot the line chart
plt.figure(figsize=(10, 6))
plt.plot(monthly_sales, marker='o', color='y', linestyle='-', linewidth=2)


# Add titles and labels
plt.title('Monthly Sales Trend', fontsize=14)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Sales ($)', fontsize=12)
plt.grid(True)

# Show the plot
plt.xticks(rotation=45)
plt.tight_layout()  # To avoid overlap
plt.show()
```

### Bar chart for Top Products

```python
import matplotlib.pyplot as plt

product = most_purchased_sorted['PRODUCTLINE']
quantity = most_purchased_sorted['QUANTITYORDERED']

# Create the bar chart
plt.figure(figsize=(10, 6))
plt.bar(product,quantity, color='skyblue')
plt.title('Top Products', fontsize=16)
plt.xlabel('PRODUCT', fontsize=14)
plt.ylabel('QUANTITY', fontsize=14)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Show the plot
plt.tight_layout()
plt.show()
```
